*TOP X TEAM AGAINST WHOM Y PLAYER SCORED MOST RUNS*

```python
# let's genaralise this for every player using function
def runs_against_team(batsman_name,X):
 target=delivery[delivery['batsman']==batsman_name]
 return target.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(X)
runs_against_team('MS Dhoni',4)
 
```

**isin() FUNCTION**

*Ques. WHO IS THE MOST DESTRUCTIVE BATSMAN IN DEATH OVERS ON BASIS OF STRIKE RATE*<br>
 *STRIKE RATE=(NUMBER OF RUNS/NUMBER OF BALLS)x100*<br>
*MINIMUM 100 BALL FACED IN DEATH OVERS i.e 16-20*<br>

```python
# let's find how many balls played by each batsman
ball=delivery.groupby('batsman').size()
ball.sort_values(ascending=False).head()
```

```python
delivery.head(2)
```

```python
maski=delivery['over']>=16
ove=delivery[maski]
# we slice the data of overs between 16-20
```

```python
ball_faced=ove.groupby('batsman').size()
ball_100=ball_faced[ball_faced>100]
ball_100.shape
# players faced>100 balls i.e 99 players 
```

```python
batsman_list=ball_100.index.tolist()
# list of all the batsman
```

```python
final=ove[ove['batsman'].isin(batsman_list)]
# store the data of those players who are in batsman_list
```

```python
runs=final.groupby('batsman')['batsman_runs'].sum()
runs
```

```python
balls=final.groupby('batsman').size()
balls
```

```python
sr=(runs/balls)*100
sr
```

**MERGING**

*QUES. FIND THE ORANGE CAP HOLDER OF EACH SEASON*

```python
new=delivery.merge(data,left_on='match_id',right_on='id') # match_id and id is common 
new
```

```python
new.groupby(['season','batsman'])['batsman_runs'].sum() # series with a multiple index
# we changed it to dataframe by using reset.index()
```

```python
new.groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False)
```

```python
new.groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index()
# we have the data of top run getters of all season now we just use drop duplicates to find 
# one player from each season i.e orange cap holder
```

```python
ans=new.groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index().drop_duplicates(subset='season',keep='first')
ans
```
