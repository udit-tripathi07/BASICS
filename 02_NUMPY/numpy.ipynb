NUMPY vs BASIC PYTHON

```python
""" To give you an idea of the performance difference, consider a NumPy array of one
million integers, and the equivalent Python list """

my_arr=np.arange(1_000_000)       # numpy array
my_list=list(range(1_000_000))    # python list
```

```python
import sys
```

```python
print(sys.getsizeof(7)*len(my_list))
```

```python
print(my_arr.itemsize*my_arr.size) # takes less space than ordinary python libraries
```

```python
import time
my_list2=range(1_000_000)
start_time=time.time()                       
my_list2 = [x * 2 for x in my_list]  # multiply each element by 2
print(time.time()-start_time)        # 0.07365655899047852 ms
```

```python
my_arr2=np.arange(1_000_000)
start_time=time.time()
my_arr2=my_arr2*2
print(time.time()-start_time)  # takes very much less time than python list 0.003305673599243164ms
```
