```python
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
iris=sns.load_dataset('iris')
```

**CATEGORICAL SCATTER PLOTS**

```python
# strip plot
sns.stripplot(data=tips,y='total_bill',x='day')
```

```python
sns.stripplot(data=tips,y='total_bill',x='day',jitter=False)
```

```python
# catplot -> figure level function like displot
sns.catplot(data=tips,x='day',y='total_bill',kind='strip',jitter=0.01,hue='sex')
```

```python
# swarmplot

sns.catplot(data=tips,x='day',y='total_bill',kind='swarm',hue='sex')
```

**CATEGORICAL DISTRIBUTION PLOT** 

*BOX PLOT*

```python
# box plot
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
sns.catplot(data=tips,x='sex',y='total_bill',kind='box')
```

```python
sns.boxplot(data=tips,y='total_bill',hue='sex')
```

*VIOLIN PLOT (BOXPLOT+KDEPLOT)*

```python
sns.violinplot(data=tips,x='day',y='total_bill',hue='sex')
```

**CATEGORICAL ESTIMATE PLOT -> FOR CENTRAL TENDENCY**

*BARPLOT*

```python
sns.barplot(data=tips,x='sex',y='total_bill')
```

```python
import numpy as np
sns.barplot(data=tips,x='sex',y='total_bill',hue='smoker',estimator=np.min)
```

```python
sns.barplot(data=tips,x='sex',y='total_bill',hue='smoker',ci=None)
# remove the errorbar
```

*POINT PLOT*

```python
sns.pointplot(data=tips,x='sex',y='total_bill',hue='smoker',ci=None)
# connects the bar
```

*COUNTPLOT*

```python
sns.countplot(data=tips,x='sex',hue='day')
```

```python
# FACETING USING CATPLOT
sns.catplot(data=tips,x='sex',hue='day',col='smoker',kind='count')
```

**REGRESSION PLOT**

```python
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
# regplot-> axes level function
# so hue function is not available here, so we use lmplot
sns.regplot(data=tips,x='total_bill',y='tip',color='red')
```

```python
#lmplot-> figure level function k
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')
```

**MULTI PLOT**

*ANOTHER WAY TO PLOT FACETGRID*

```python
# figure level -> relplot, displot, catplot,lmplot
# we study this
sns.catplot(data=tips,x='sex',y='total_bill',kind='violin',col='day',row='time')
```

```python
g=sns.FacetGrid(data=tips,col='day',row='time')
g
# creates a empty grid now you can plot whatever graph you want
```

```python
g=sns.FacetGrid(data=tips,col='day',row='time')
g.map(sns.violinplot,'sex','total_bill')
```

**PAIRWISE PLOTTING**

*PAIRPLOT*

```python
sns.pairplot(iris)
```

```python
sns.pairplot(iris,hue='species')
```

```python
sns.pairplot(tips,hue='sex')
```

*PAIRGRID*

```python
g=sns.PairGrid(iris)
g
# creates a grid
```

```python
g=sns.PairGrid(iris,hue='species')
g.map(sns.scatterplot)
```

```python
# diag and offdiag
g=sns.PairGrid(iris,hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
# map_upper and map_lower
g=sns.PairGrid(iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

**JOINTGRID & JOINTPLOT**

```python
sns.jointplot(data=tips,x='total_bill',y='tip',kind='reg')
```

```python
g=sns.JointGrid(data=tips,x='total_bill',y='tip',hue='sex')
g.plot(sns.scatterplot,sns.violinplot)
# draw whichever we want
```
