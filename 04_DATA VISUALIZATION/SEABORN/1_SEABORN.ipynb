**1.RELATIONAL PLOT**

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
sns.set(rc={"figure.figsize": (5, 5)})
```

```python
tips =sns.load_dataset('tips')
tips.head()
```

```python
# scatter plot -> axes level
plt.figure(figsize=(3.5,3.5))
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
# relplot -> figure level (square shape)
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter')
# we can also plot lineplot using this by just changing the kind function 
```

```python
plt.figure(figsize=(3.5,3.5))
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex')
# scatter plot -> axes level
```

```python
plt.figure(figsize=(5,5))
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
# hue and style for more info
```

```python
# lineplot
gap=px.data.gapminder()
gap.head()
ind=gap[gap['country']=='India']
ind
```

```python
sns.lineplot(data=ind,x='year',y='gdpPercap')
```

```python
# using relplot
sns.relplot(data=ind,x='year',y='gdpPercap',kind='line')
```

```python
ipc=gap[gap['country'].isin(['India','Pakistan','China'])]
```

```python
sns.relplot(data=ipc,x='year',y='gdpPercap',kind='line',hue='country')
```

```python
sns.lineplot(data=ipc,x='year',y='gdpPercap',hue='country')
```

YOU CAN SEE IN ABOVE GRAPHS THAT RELPLOT WRITE THE OTHER INFO OUTSIDE THE GRAPH SINCE IT IS GRAPH LEVEL WHEREAS LINEPLOT WRITE IT INSIDE The GRAPH SINCE IT IS A AXIS LEVEL FUNCTION.

```python
# facet plot
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',col='sex')
# col and row is used for multiple plot
```

```python
# facet plot
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',col='sex',row='smoker')
```

```python
sns.relplot(data=gap,x='pop',y='lifeExp',col='year',col_wrap=4)
```

**2.DISTRIBUTION PLOT**

```python
sns.histplot(data=tips,x='total_bill')
# tips.head()
```

```python
sns.histplot(data=tips,x='day')
```

```python
sns.histplot(data=tips,x='total_bill',hue='sex')
```

```python
# kdeplot
sns.kdeplot(data=tips,x='total_bill')
```

```python
# rugplot 
# used with other plots to show the graph in different manner
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

**#HEATMAP**

```python
temp=gap.pivot(index='country',columns='year',values='lifeExp')
temp.head()
```

```python
sns.heatmap(temp)
```

```python
euro=gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
euro.head()
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(euro,annot=True,linewidth=0.5)
```
