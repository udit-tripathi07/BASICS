***ADVANCED MATPLOTLIB***

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris=pd.read_csv('iris.csv')
iris.sample(5)
```

```python
# we replace the name of species with numeric value for coloring 
iris['Species']=iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.head(5)
```

**COLORING**

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])
plt.colorbar()
```

```python
plt.figure(figsize=(5,3))  # size of the graph
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])
plt.colorbar()
```

**ANOTATIONS**

```python
batters=pd.read_csv('batter.csv')
batters.head()
```

```python
samp=batters.head(100).sample(25,random_state=5)
samp
```

```python
plt.scatter(samp['avg'],samp['strike_rate'])
```

```python
plt.figure(figsize=(18,11))
plt.scatter(samp['avg'],samp['strike_rate'])
for i in range(samp.shape[0]):
 plt.text(samp['avg'].values[i],samp['strike_rate'].values[i],samp['batter'].values[i])

 # syntax= plt.text(x,y,name of point)
```

```python
plt.figure(figsize=(18,11))
plt.scatter(samp['avg'],samp['strike_rate'],samp['runs']) # size of point depend on runs
for i in range(samp.shape[0]):
 plt.text(samp['avg'].values[i],samp['strike_rate'].values[i],samp['batter'].values[i])

 # syntax= plt.text(x,y,name of point)
```

**ADDING LINES ON GRAPH**

```python
plt.figure(figsize=(18,11))
plt.scatter(samp['avg'],samp['strike_rate'])
for i in range(samp.shape[0]):
 plt.text(samp['avg'].values[i],samp['strike_rate'].values[i],samp['batter'].values[i])

plt.axhline(127,color='red')
plt.axvline(30,color='red')

```

**SUBPLOTS**

```python
# other way to plot the graph
fig,ax=plt.subplots()
ax.scatter(batters['avg'],batters['strike_rate'])
ax.set_title('avg vs sr')
ax.set_xlabel('ang')
ax.set_ylabel('strike rate')
```

```python
fig,ax=plt.subplots(nrows=2,ncols=1)
# ax stores the info of all the subgraphs
```

```python
fig,ax=plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(6,6)) # shareX
ax[0].scatter(batters['avg'],batters['strike_rate'],color='red',marker='+')
ax[0].set_title('avg vs strike rate')
# ax[0].set_xlabel('avg')
ax[0].set_ylabel('strike rate')

ax[1].scatter(batters['avg'],batters['runs'],color='blue')
ax[1].set_title('avg vs runs ')
ax[1].set_xlabel('avg')
ax[1].set_ylabel('runs')
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2) 
ax[0][0].scatter(batters['avg'],batters['strike_rate'],color='red',marker='+')


ax[1][0].hist(batters['avg'],color='red')

ax[0][1].scatter(batters['avg'],batters['runs'],color='blue')

ax[1][1].hist(batters['runs'],color='blue')


```

**3D SURFACE PLOT**

```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
```

```python
xx,yy=np.meshgrid(x,y) # grid of 100 X 100
```

```python
z=xx**2+yy**2
```

```python
ax=plt.subplot(projection='3d')
ax.plot_surface(xx,yy,z)
```

```python
z=np.sin(xx)+np.sin(yy)
ax=plt.subplot(projection='3d')
ax.plot_surface(xx,yy,z,cmap='viridis')
```

**CONTOUR GRAPH**

```python
z=(xx)**2+(yy)**2
ax=plt.subplot()
ax.contour(xx,yy,z)
```

```python
z=(xx)**2+(yy)**2
ax=plt.subplot()
ax.contourf(xx,yy,z)
```

```python
z=np.sin(xx)+np.sin(yy)
ax=plt.subplot()
ax.contourf(xx,yy,z,cmap='viridis')
```

**HEATMAP**

```python
delivery=pd.read_csv('IPL_Ball_by_Ball_2022.csv')
delivery.head()
```

```python
six=delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6]))&(delivery['batsman_run']==6)]
six.head()
```

```python
grid=six.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')
grid
```

```python
plt.figure(figsize=(36,5))
plt.imshow(grid)
plt.colorbar()
```
